```python
def fix_time_seconds(text):
    if pd.isna(text):
        return text
    text = str(text)

    # Only touch text that contains AM/PM (ignore "Upon Booking", etc.)
    if re.search(r'\bAM\b|\bPM\b', text):
        # Remove any double :00 like 8:30:00:00 → 8:30
        text = re.sub(r'(\d{1,2}:\d{2})(:00)+', r'\1', text)

        # Remove :00 for whole hours like 8:00 → 8:00 (keep one only)
        text = re.sub(r'(\d{1,2}):00(?=\s*[AP]M)', r'\1:00', text)

        # Fix any trailing extra colons accidentally formed
        text = re.sub(r':+', ':', text)

    return text

df['operating hours'] = df['operating hours'].apply(fix_time_seconds)

print("✅ Seconds removed and time formats perfected!")
df['operating hours'].head(10)
```

```python
import re
import numpy as np

# Function to extract opening and closing times
def split_hours(value):
    if isinstance(value, str):
        # Handle special text cases
        if "upon" in value.lower() or "temporary" in value.lower() or "closed" in value.lower():
            return pd.Series([np.nan, np.nan])
        
        # Match "9:00 AM – 5:00 PM" pattern
        match = re.match(r"(\d{1,2}:\d{2}(?::\d{2})?\s?(?:AM|PM))\s*[–-]\s*(\d{1,2}:\d{2}(?::\d{2})?\s?(?:AM|PM))", value, re.IGNORECASE)
        if match:
            return pd.Series(match.groups())
    
    return pd.Series([np.nan, np.nan])

# Apply function to your column
df[['opening_time', 'closing_time']] = df['operating hours'].apply(split_hours)

# Drop seconds from the new columns for a cleaner look
df['opening_time'] = df['opening_time'].str.replace(r":00\s", " ", regex=True)
df['closing_time'] = df['closing_time'].str.replace(r":00\s", " ", regex=True)

# Display sample
df[['operating hours', 'opening_time', 'closing_time']].head(10)
```

```python
import re

# Helper function to standardize days
def clean_days(text):
    if pd.isna(text) or text.strip() == "" or text.lower() == "none":
        return "Not Mentioned"
    
    text = text.strip().lower()
    
    # Handle special terms
    if "daily" in text or "everyday" in text:
        return "Daily"
    if "upon booking" in text:
        return "Upon Booking"
    if "temporary" in text:
        return "Temporary Closed"
    if "season" in text:
        return "Seasonal"
    if "annual" in text:
        return "Annual Event"
    
    # Replace symbols and normalize
    text = text.replace("–", "-").replace(",", " ").replace(";", " ")
    text = re.sub(r"\s+", " ", text).strip()
    
    # Map short forms to full names
    replacements = {
        "mon": "Monday", "tue": "Tuesday", "wed": "Wednesday",
        "thu": "Thursday", "fri": "Friday", "sat": "Saturday", "sun": "Sunday"
    }
    
    for short, full in replacements.items():
        text = re.sub(rf"\b{short}\b", full, text)
    
    # Normalize ranges (e.g., Monday - Friday)
    text = text.title()
    text = text.replace(" - ", " – ")
    
    return text

# Apply cleaning
df['operating days'] = df['operating days'].apply(clean_days)
df['closing days'] = df['closing days'].apply(clean_days)

# Preview result
df[['operating days', 'closing days']].head(15)
```

```python
import re

# Define mapping for short forms
day_map = {
    "Mon": "Monday",
    "Tue": "Tuesday",
    "Wed": "Wednesday",
    "Thu": "Thursday",
    "Thur": "Thursday",
    "Fri": "Friday",
    "Sat": "Saturday",
    "Sun": "Sunday"
}
```

```python
def standardize_days(text):
    if pd.isna(text) or text.strip().lower() in ['none', 'not mentioned', 'n/a']:
        return "Not mentioned"
    
    text = text.strip()

    # Handle special cases
    lower_text = text.lower()
    if "upon booking" in lower_text:
        return "Upon Booking"
    if "temporary closed" in lower_text or "season" in lower_text:
        return "Temporarily Closed"
    if "annual" in lower_text or "festival" in lower_text:
        return "Annual Event"

    # Replace short day names with full names
    for short, full in day_map.items():
        text = re.sub(rf"\b{short}\b", full, text, flags=re.IGNORECASE)
    
    # Replace common separators with a consistent " – "
    text = re.sub(r"[-–,]", " – ", text)
    text = re.sub(r"\s+", " ", text).strip()  # remove double spaces

    return text
```

```python
df['operating days'] = df['operating days'].apply(standardize_days)
df['closing days'] = df['closing days'].apply(standardize_days)
```

```python
def final_day_cleanup(text):
    if pd.isna(text) or text.lower() in ['not mentioned', 'n/a']:
        return "Not mentioned"

    text = text.strip()

    # Normalize general terms
    replacements = {
        "Weekend": "Saturday – Sunday",
        "Weekends": "Saturday – Sunday",
        "Weekdays": "Monday – Friday",
        "Daily": "Daily",
        "Annual Event": "Annual Event",
        "Temporarily Closed": "Temporarily Closed",
        "Upon Booking": "Upon Booking"
    }
    for old, new in replacements.items():
        if old.lower() in text.lower():
            return new

    # Fix day pair formats (e.g. "Saturday Sunday" -> "Saturday – Sunday")
    text = re.sub(r"(?<=Saturday)\s+(?=Sunday)", " – ", text, flags=re.IGNORECASE)
    text = re.sub(r"(?<=Friday)\s+(?=Saturday)", " – ", text, flags=re.IGNORECASE)
    text = re.sub(r"(?<=Tuesday)\s+(?=Friday)", " – ", text, flags=re.IGNORECASE)
    text = re.sub(r"(?<=Monday)\s+(?=Tuesday)", " – ", text, flags=re.IGNORECASE)

    # Remove duplicate spaces and ensure consistent separator
    text = re.sub(r"\s*[-–]\s*", " – ", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text

# Apply to both columns
df['operating days'] = df['operating days'].apply(final_day_cleanup)
df['closing days'] = df['closing days'].apply(final_day_cleanup)
```

```python
# Step 1: Find rows where min price > max price
invalid_prices = df[df['min price (RM)'] > df['max price (RM)']]
print("⚠️ Invalid price ranges detected:")
display(invalid_prices)

# Step 2: Fix cases where min > max (swap them)
df.loc[df['min price (RM)'] > df['max price (RM)'], ['min price (RM)', 'max price (RM)']] = \
    df.loc[df['min price (RM)'] > df['max price (RM)'], ['max price (RM)', 'min price (RM)']].values

# Step 3: Identify zero or negative prices
invalid_zero = df[(df['min price (RM)'] <= 0) | (df['max price (RM)'] <= 0)]
print("⚠️ Zero or negative prices detected:")
display(invalid_zero)
```

```python
def classify_price(row):
    if row['min price (RM)'] == 0 or row['max price (RM)'] == 0:
        return "Free / Not mentioned"
    elif row['min price (RM)'] < 0 or row['max price (RM)'] < 0:
        return "Invalid"
    else:
        return "Paid"

df['price_type'] = df.apply(classify_price, axis=1)
```
